# Aritmética de Ponto Flutuante (APF)

## Introdução
A Aritmética de Ponto Flutuante é um sistema adotado para representar números reais em sistemas computacionais. Como computadores possuem recursos de hardware limitados, eles podem representar apenas um subconjunto finito $F \subset \mathbb{R}$ de todos os números reais.

## Representação de um número em APF
No sistema da APF, um computador representa um número $x$ como:

$$x = (-1)^s \times (0.d_1d_2\ldots d_t)_\beta \times \beta^e$$

## Componentes
Cada elemento da representação acima possui um significado específico:

- **$s \in \{0,1\}$**: sinal do número (0 para positivo, 1 para negativo)
- **$\beta$**: inteiro $\geq 2$, denominado **base** do sistema (geralmente 2, 10 ou 16)
- **$m = 0.d_1d_2\ldots d_t$**: a **mantissa** ou **significando**, composto por $t$ dígitos $d_i$ onde $0 \leq d_i \leq (\beta - 1)$ 
  - **Nota importante:** $d_1 \neq 0$ garante que a representação seja única (é o requisito de **normalização**)
- **$e$**: expoente inteiro, limitado por $l \leq e \leq u$, que define a magnitude e a posição do ponto decimal

## Conjunto e Sistema de Ponto Flutuante
Os números representados nesta forma definem o subconjunto de ponto flutuante:

$$F = F(\beta, t, l, u) \subset \mathbb{R}$$

onde os parâmetros do sistema são: base $\beta$, precisão (número de dígitos) $t$, expoente mínimo $l$ e expoente máximo $u$.

## Números representáveis extremos
O menor e o maior número positivo representável em um sistema $F(\beta, t, l, u)$ são:

$$x_{\min} = (0.10\ldots0)_\beta \times \beta^l = \beta^{l-1}$$

$$x_{\max} = (0.\underbrace{(\beta-1)(\beta-1)\ldots(\beta-1)}_{t \text{ dígitos}})_\beta \times \beta^u = (1 - \beta^{-t})\beta^u$$

Estes limites definem o intervalo de números que podem ser representados exatamente no sistema de ponto flutuante.

### Experimento

\begin{equation} \frac{(1 + x) - 1}{x} \end{equation}

Para números reais, esta expressão é identicamente igual a 1 para qualquer
`x` não-nulo. No entanto, quando trabalhamos com um sistema de ponto flutuante
como o discutido nas seções anteriores – com base, mantissa normalizada e
expoente limitado – a aritmética não é exata.

A precisão finita implica que, à medida que `x` se torna pequeno em relação a
`1`, o resultado de `1 + x` pode ser arredondado de volta para `1` porque `x`
não é representável dentro da mantissa. A subtração `(1+x) - 1` então produz
zero e o quociente falha em reproduzir o valor esperado. Esse comportamento é
uma manifestação prática da perda de significância e dos limites de
representabilidade que derivam do sistema $F(\beta,t,l,u)$ descrito acima.

O experimento ilustra, de forma simples, como erros de arredondamento e a
normalização dos números podem afetar cálculos que, em teoria pura, seriam
triviais.

```python
def f(x):
    return ((1+x)-1)/x

f(1), f(1e-5), f(1e-10), f(1e-15), f(1e-30)
```

### Epsilon da Máquina

O **epsilon da máquina** (ou **machine epsilon**), denotado comumente como $\epsilon_m$ ou $\text{eps}$, é a diferença relativa positiva mais pequena representável em um sistema de ponto flutuante. Em outras palavras, é o menor número positivo $\epsilon$ tal que:

$$1 + \epsilon \neq 1$$

em aritmética de ponto flutuante.

#### Significado

Este valor representa a **precisão máxima** que o computador consegue alcançar. Qualquer número menor que $\epsilon_m$ não será distinguível de zero quando somado a 1, devido aos limites de representação impostos pela mantissa e pela normalização.

Matematicamente, epsilon da máquina pode ser expresso como:

$$\epsilon_m = \beta^{1-t}$$

onde $\beta$ é a base e $t$ é o número de dígitos da mantissa.

#### Importância Prática

O epsilon da máquina fornece um limite fundamental para:
- **Erros de arredondamento**: qualquer operação aritmética terá um erro relativo da ordem de $\epsilon_m$
- **Perda de significância**: quando dois números próximos são subtraídos, os dígitos significativos podem ser perdidos
- **Tolerâncias numéricas**: serve como base para definir quando dois números devem ser considerados iguais em comparações

O algoritmo que vem na célula abaixo computa realmente o epsilon da máquina no seu ambos de execução.

```python
epsilon = 1
while 1 + epsilon != 1:
    epsilon /= 2
epsilon = 2*epsilon
epsilon
```
