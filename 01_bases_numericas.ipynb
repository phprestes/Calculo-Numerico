# Conversão de binário para decimal — 2 em evidência (forma em cascata)

Uma forma de evidenciar o 2 é fatorar iterativamente os termos (método de Horner). Para o número binário $b_n\dots b_0$:

$$ (b_n b_{n-1} \dots b_1 b_0)_2 = b_n \cdot 2^n + b_{n-1} \cdot 2^{n-1} + \dots + b_1 \cdot 2 + b_0 $$

Ou equivalentemente, fatorando 2:

$$ (b_n b_{n-1} \dots b_1 b_0)_2 = 2(b_n \cdot 2^{n-1} + b_{n-1} \cdot 2^{n-2} + \dots + b_1) + b_0 $$

Aplicando Horner iterativamente:

$$ (b_n b_{n-1} \dots b_1 b_0)_2 = 2(\ldots 2(2(b_n) + b_{n-1}) + \ldots + b_1) + b_0 $$

Exemplo — converter $(10111)_2$:

Avaliação passo a passo (da esquerda para direita):
- $1$
- $1 \cdot 2 + 0 = 2$
- $2 \cdot 2 + 1 = 5$
- $5 \cdot 2 + 1 = 11$
- $11 \cdot 2 + 1 = 23$

Portanto, $(10111)_2 = 23_{10}$. O método de Horner evidencia como multiplicamos por 2 a cada passo e adicionamos o próximo dígito, facilitando o cálculo sequencial.

```python
def binario_decimal(numero_binario):
    res = 0
    for n in numero_binario:
        if n == "1":
            res = 2*res + 1
        elif n == "0":
            res = 2*res
        else:
            break
    return res

# binario_decimal("10111")
binario_decimal("110101011")
```

# Conversão de decimal para binário — método dos restos

Para converter um número decimal para binário, divide-se repetidamente o número por 2 e anota-se o resto de cada divisão. Os restos, lidos de baixo para cima (do último para o primeiro), formam o número binário.

Passos:
1. Divida o número decimal por 2.
2. Anote o resto (0 ou 1).
3. Use o quociente como novo número e repita enquanto o quociente for maior que 0.
4. Leia os restos na ordem inversa (do último resto obtido para o primeiro) — esse é o número em base 2.

Exemplo: converter $23_{10}$ para binário.

- $23 \div 2 = 11$ quociente, resto $1$.
- $11 \div 2 = 5$ quociente, resto $1$.
- $5 \div 2 = 2$ quociente, resto $1$.
- $2 \div 2 = 1$ quociente, resto $0$.
- $1 \div 2 = 0$ quociente, resto $1$.

Resto (do último para o primeiro): $1\;0\;1\;1\;1$, portanto $23_{10} = 10111_2$.

Observação: este método produz diretamente os bits do menos significativo para o mais significativo; por isso a necessidade de inverter a sequência de restos ao final.

```python
def decimal_binario(numero_decimal):
    res = ""
    while numero_decimal > 0:
        res = str(numero_decimal % 2) + res
        numero_decimal //= 2
    return res

decimal_binario(70)
    
```

# Conversão de decimal fracionário para binário — método de multiplicação sucessiva por 2

Para converter um número decimal com parte fracionária para binário, utiliza-se o método de multiplicação sucessiva por 2. Este método é o inverso do processo de conversão de binário fracionário para decimal.

## Método
Os passos para converter a parte fracionária são:

1. Multiplique o número fracionário por 2.
2. Se o resultado for **maior ou igual a 1**, o bit é **1**; caso contrário, o bit é **0**.
3. Extraia a parte inteira (o bit obtido).
4. Continue com a parte fracionária do resultado (subtraia a parte inteira se o resultado foi ≥ 1).
5. Repita os passos 1–4 até encontrar um resultado nulo ou até atingir a precisão desejada (muitas frações decimais resultam em dízimas infinitas em binário).

## Fórmula
Para converter $0.d$ (fração decimal) para binário:

$$0.d_{10} = 0.b_1b_2b_3\ldots_2$$

onde cada $b_i \in \{0, 1\}$ é obtido através da multiplicação sucessiva por 2.

## Exemplo
Converter $0.625_{10}$ para binário:

- $0.625 \times 2 = 1.25$ → bit **1**, restante $0.25$
- $0.25 \times 2 = 0.5$ → bit **0**, restante $0.5$
- $0.5 \times 2 = 1.0$ → bit **1**, restante $0.0$

Resultado: $0.625_{10} = 0.101_2$ (conversão exata).

## Observação
Nem todas as frações decimais terminam em binário. Por exemplo, $0.1_{10}$ resulta em uma dízima infinita $0.0001100110011\ldots_2$. Por isso, na prática, estabelece-se um limite de casas (precisão) para a conversão, evitando loops infinitos.

```python
def decimal_binario_fracionario(numero_decimal):
    print("0.", end="")
    numero_decimal = 0.1
    limite_casas = 20  # Trava de segurança para evitar loop infinito

    while numero_decimal != 0 and limite_casas > 0:
        numero_decimal *= 2
        d = int(numero_decimal) 
        print(d, end="")
        numero_decimal -= d 
        limite_casas -= 1
```

# Conversão de binário fracionário para decimal

Para converter um número binário com parte fracionária para decimal, separamos a conversão em duas partes:

## Parte inteira
A parte inteira segue o método de Horner já apresentado, multiplicando iterativamente por 2.

## Parte fracionária
Para a parte fracionária, cada dígito binário $b_i$ na posição $i$ após o ponto representa $b_i \cdot 2^{-i}$:

$$0.b_1b_2b_3\ldots = b_1 \cdot 2^{-1} + b_2 \cdot 2^{-2} + b_3 \cdot 2^{-3} + \ldots$$

**Método iterativo:** multiplicamos a fração por 10 (em decimal), extraímos a parte inteira do resultado (que forma os dígitos decimais), e repetimos com a parte fracionária restante.

## Exemplo
Para converter $(110101011.111111101)_2$:
- Parte inteira: $(110101011)_2 = 427_{10}$
- Parte fracionária: $(0.111111101)_2$ convertida iterativamente

Resultado: $(110101011.111111101)_2 \approx 427.9980..._10$

```python
def multiplicar_binario_por_dez(numero_binario):
    # Proteção: se o número não tiver ponto, adicionamos ".0"
    if "." not in numero_binario:
        numero_binario += ".0"
        
    _, pos_virgula = numero_binario.split('.')
    casas = len(pos_virgula)
    
    # Removemos o ponto para trabalhar apenas com inteiros
    numero_binario_inteiro = numero_binario.replace(".", "")
    
    # Converte, multiplica por 10 decimal, e volta para binário
    resultado_inteiro_dec = binario_decimal(numero_binario_inteiro) * 10
    resultado_binario = decimal_binario(resultado_inteiro_dec)

    # Recoloca o ponto na posição correta
    resultado_final = "." + resultado_binario[-casas:]
    if resultado_binario[:-casas] == "":
        resultado_final = "0" + resultado_final
    else:
        resultado_final = resultado_binario[:-casas] + resultado_final
    
    return resultado_final.rstrip("0").rstrip(".")  # Remove zeros à direita e o ponto se não houver parte decimal

def binario_decimal_fracao(numero_binario):
    if "." in numero_binario:
        parte_int_bin, parte_frac_bin = numero_binario.split(".")
    else:
        parte_int_bin, parte_frac_bin = numero_binario, ""
    
    res = str(binario_decimal(parte_int_bin))
    if not parte_frac_bin:
        return res
    
    res += "."
    fracao_atual = "0." + parte_frac_bin

    while True:
        w = multiplicar_binario_por_dez(fracao_atual)
        if '.' not in w:
            return res + str(binario_decimal(w))
        z, fracao_atual = w.split(".")
        fracao_atual = "0." + fracao_atual
        res += str(binario_decimal(z))

binario_decimal_fracao("110101011.111111101")
```

## Imprecisão na representação binária de decimais

Embora tenhamos visto métodos precisos para converter entre bases, nem todas as frações decimais podem ser representadas exatamente em binário. É ainda mais relevante em aritmética de ponto flutuante, onde a mantissa tem um número finito de dígitos.

### O problema com 0.1 em binário

O número decimal $0.1_{10}$ é uma fração que, quando convertida para binário usando o method de multiplicação sucessiva por 2, resulta em uma **dízima infinita não-periódica** (ou praticamente infinita):

$$0.1_{10} = 0.\overline{0001100110011...}_2$$

Isto significa que **não existe representação exata** de $0.1$ em binário com um número finito de dígitos. O computador deve, então, arredondar a representação a um número limitado de casas, introduzindo um pequeno erro em cada operação.

### Acumulação de erros

Quando realizamos operações repetidas com números que não podem ser representados exatamente, os erros de arredondamento **se acumulam**. Por exemplo, quando somamos $0.1$ uma mil vezes:

$$\sum_{i=1}^{1000} 0.1$$

Teoricamente, o resultado deveria ser exatamente $100.0$. Porém, cada soma introduz um pequeno erro devido à representação imprecisa de $0.1$. Após mil iterações, esses erros pequenos se combinam, resultando em um desvio perceptível do valor esperado.

```python
n = 0
for i in range(0, 1000):
    n += 0.1
n
```
