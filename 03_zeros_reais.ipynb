# Zeros Reais de Funções: Métodos Iterativos

## Introdução
A busca por zeros (ou raízes) de funções reais é um problema fundamental em cálculo numérico. O objetivo é encontrar o(s) valor(es) de $x$ tal que $f(x) = 0$. Como nem sempre é possível obter uma solução analítica, recorremos a **métodos iterativos** que constroem uma sequência de aproximações cada vez mais precisas.

## Método da Bissecção
O método da bissecção é um algoritmo de busca de raízes concebido a partir do Teorema do Valor Intermediário (Bolzano). Se uma função contínua $f(x)$ possui sinais opostos nos extremos de um intervalo $[a, b]$, ou seja, $f(a) \cdot f(b) < 0$, então existe pelo menos um zero de $f$ dentro deste intervalo.

## Componentes do Algoritmo
O método divide repetidamente o intervalo subjacente pela metade e seleciona o subintervalo que contém a raiz. Os elementos principais são:

- **$[a, b]$**: Intervalo inicial tal que $f(a) \cdot f(b) < 0$
- **$x_k$**: Ponto médio do intervalo avaliado a cada iteração, dado por $x_k = \frac{a + b}{2}$
- **$\epsilon$**: Tolerância desejada para o critério de parada, representando o erro admissível
- **$N$**: Número máximo de iterações permitidas caso a convergência demore

Em cada iteração $k$, avalia-se a função no ponto médio $f(x_k)$. Se $f(a) \cdot f(x_k) < 0$, a raiz está no subintervalo inferior $[a, x_k]$. Caso contrário, está no subintervalo superior $[x_k, b]$. O processo se repete até que o tamanho do intervalo $|b - a|$ atenda à tolerância $\epsilon$ ou $f(x_k) = 0$.

## Características do Método
- **Convergência Incondicional**: É um método globalmente convergente. Isso significa que, partindo de um intervalo válido (com mudança de sinal), ele sempre encontrará uma solução.
- **Velocidade (Taxa)**: É considerado lento. Sua taxa de convergência é **linear** e isola a raiz diminuindo o tamanho do intervalo pela metade a cada passo.

O erro absoluto máximo admissível na $k$-ésima iteração é estritamente limitado por:

$$E_k \leq \frac{b - a}{2^{k}}$$

### Experimento

No exemplo abaixo, implementaremos o Método da Bissecção para encontrar a raiz da função $f(x) = x^2 - 2$ no intervalo $[1, 2]$. Este cálculo busca uma aproximação para testar as condições de convergência do algoritmo.

Sendo uma função contínua onde $f(1) = -1$ e $f(2) = 2$, pelos princípios de Bolzano, existe uma raiz real neste intervalo correspondente a $\sqrt{2}$. Monitoraremos as iterações e o erro final atingido.

```python
def bisseccao(f, a, b, tol=1e-5, max_iter=100):
    if f(a) * f(b) >= 0:
        raise ValueError("O intervalo [a, b] deve conter uma raiz (f(a) e f(b) com sinais opostos).")
    
    k = 0
    while (b - a) > tol and k < max_iter:
        xk = (a + b) / 2.0
        
        if f(xk) == 0:
            return xk, k, 0.0
            
        if f(a) * f(xk) < 0:
            b = xk
        else:
            a = xk
        k += 1
        
    xk = (a + b) / 2.0
    return xk, k, abs(f(xk))

# Definindo a função f(x) = x^2 - 2
def f_exemplo(x):
    return x**2 - 2

# Buscando a raiz na tolerância 1e-6
raiz, iteracoes, erro_funcao = bisseccao(f_exemplo, 1.0, 2.0, tol=1e-6)
raiz, iteracoes, erro_funcao
```

## Método da Falsa Posição (Regula Falsi)

O método da falsa posição é uma variação mais sofisticada do método da bissecção. Ele também parte da premissa do Teorema do Valor Intermediário exigindo um intervalo $[a, b]$ com $f(a) \cdot f(b) < 0$. No entanto, em vez de simplesmente dividir o intervalo pela metade cegamente, ele considera _a magnitude_ dos valores de $f(a)$ e $f(b)$ para tentar descobrir onde a raiz está mais próxima.

A ideia principal é traçar uma reta secante ligando os pontos cartesianos $(a, f(a))$ e $(b, f(b))$. A interseção dessa reta com o eixo $x$ fornecerá o próximo palpite para a nossa raiz.

## Componentes do Algoritmo

Os elementos principais são muito parecidos com os do método da bissecção, mas a forma como $x_k$ é escolhido muda.

- **$[a, b]$**: Intervalo de busca que confina a raiz em sua respectiva iteração
- **$x_k$**: Ponto de interseção da secante com o eixo das abscissas. Pode ser explicitamente formulado como:
  
  $$x_k = \frac{a \cdot f(b) - b \cdot f(a)}{f(b) - f(a)}$$

- **Critérios de Parada**: Como o tamanho do intervalo nem sempre tende a zero (um dos limites pode se fixar), o controle de tolerância $\epsilon$ se foca puramente no erro avaliado na imagem e/ou diferença entre iterações da raiz: $|f(xk)| < \epsilon$ ou $|x_k - x_{k-1}| < \epsilon$.

Seguindo o processo, estipula-se o novo intervalo da mesma forma que na bissecção usando a análise de sinais para preservar a convergência. Se $f(a) \cdot f(x_k) < 0$, o novo limite superior é estipulado sendo $b = x_k$. Senão, $a = x_k$.

## Características do Método
- **Garantia de Convergência**: Como exige que sempre haja uma mudança de sinal garantida pelo intervalo confinado, a Falsa Posição também é incondicionalmente convergente.
- **Velocidade**: Geralmente converge mais rápido do que a bissecção, possuindo uma taxa de convergência tipicamente aproximada à linear (e sob certas condições otimizadas, chega a apresentar perfil superlinear aproximado de 1.618 em algumas definições modificadas).
- **Risco de Ancoramento (Stalling)**: Em funções estritamente convexas ou côncavas ao redor da raiz, um dos limites pode nunca mudar. A secante ficará "ancorada", o que faz o intervalo se reduzir de um lado só.


### Experimento

Analogamente à bissecção, aplicaremos o método da Falsa Posição para desvendar a raiz de $f(x) = x^2 - 2$ confinada no intervalo $[1, 2]$ e observar o quão rápido chegamos a um resultado numérico que traga erro inferior à tolerância estipulada.

Veremos que as iterações geralmente exigem menos passadas (loops) para abater a tolerância do que a bissecção.


```python
def posicao_falsa(f, a, b, tol=1e-5, tol_val=1e-5, max_iter=100):
    if f(a) * f(b) >= 0:
        raise ValueError("O intervalo [a, b] deve conter uma raiz (f(a) e f(b) com sinais opostos).")

    k = 0
    while (b - a) > tol and k < max_iter:
        if abs(f(a)) < tol_val: 
            return a, k, abs(f(a))
        if abs(f(b)) < tol_val: 
            return b, k, abs(f(b))

        xk = (a * f(b) - b * f(a)) / (f(b) - f(a))
        
        if f(xk) == 0:
            return xk, k, 0.0
            
        if f(a) * f(xk) < 0:
            b = xk
        else:
            a = xk
        k += 1
        
    xk = (a * f(b) - b * f(a)) / (f(b) - f(a))
    return xk, k, abs(f(xk))

raiz, iteracoes, erro_funcao = posicao_falsa(f_exemplo, 1.0, 2.0, tol=1e-6, tol_val=1e-6)
raiz, iteracoes, erro_funcao
```

## Método do Ponto Fixo (Iterativo Linear)

O Método do Ponto Fixo diferencia-se fundamentalmente da Bissecção e da Falsa Posição, já que não trabalha confinando inteiramente a raiz através de mudanças de sinal. Trata-se de um chamado **método aberto**.

O problema fundamental de encontrar $x$ tal que $f(x) = 0$ é transformado algebricamente num problema de achar um ponto fixo $x = \Phi(x)$. A intuição aqui é que estipulamos uma iterativa direta $x_{k+1} = \Phi(x_k)$, de modo que, ao aplicá-la sucessivamente partindo de um chute inicial $x_0$, a sequência $(x_k)$ divirja ou convirja estritamente parar a raiz que buscamos.

## Componentes do Algoritmo

Os elementos agora dizem respeito não mais a um limite inferior e superior de pesquisa, mas apenas um local de partida no eixo-x.

- **$\Phi(x)$**: Função de Iteração (ou Função Passo). Precisa ser definida pelo usuário no instante de formular o problema. Por exemplo, na equação $x^2 - 2 = 0$, uma escolha válida (embora existam várias) pode ser $\Phi(x) = \frac{x}{2} + \frac{1}{x}$.
- **$x_0$**: Chute ou palpite inicial. É crucial para métodos abertos. Um palpite ruim pode levar à divergência computacional.
- **$x_{k+1}$**: O próximo valor deduzido calculando $\Phi(x_k)$.

- **Critérios de Parada**: Como antes, o controle de tolerância $\epsilon$ verifica $|f(x_k)| < \epsilon$ (para checar a imagem original) ou analisa a proximidade das iterações $|x_k - x_{k-1}| < \epsilon$.

## Características do Método
- **Garantia de Convergência Limitada**: O método do Ponto Fixo só irá convergir se a derivada da função de iteração, no entorno da raiz $\bar{x}$, possuir módulo estritamente menor que 1: $|\Phi'(\bar{x})| < 1$. Caso $|\Phi'(x)| > 1$, o método divergirá ao tentar localizar a raiz.
- **Velocidade**: A convergência do método usualmente é **Linear**, similar à Bissecção. Dependendo de termos nulos nas derivadas seguintes (como ocorre na variação do método de Newton-Raphson), a convergência pode ser acelerada para um padrão Quadrático ou superior.


### Experimento

Continuaremos nossa missão: encontrar a raiz de $f(x) = x^2 - 2 = 0$ próximo de $[1, 2]$. 
Reformularemos algelbraicamente a equação de tal modo a forjar $x = \Phi(x)$:

$$x^2 = 2 \implies 2x^2 = x^2 + 2 \implies 2x = x + \frac{2}{x} \implies x = \frac{x}{2} + \frac{1}{x}$$

Logo, $\Phi(x) = \frac{x}{2} + \frac{1}{x}$.

Usaremos o palpite inicial $x_0 = 1.5$ e verificaremos as iterações e a tolerância atingida.


```python
def ponto_fixo(phi, x0, f_original, tol=1e-5, max_iter=100):
    x_k = x0
    k = 0
    
    while k < max_iter:
        erro = abs(f_original(x_k))
        if erro < tol:
            return x_k, k, erro
            
        x_k_next = phi(x_k)
        
        if abs(x_k_next - x_k) < tol:
            return x_k_next, k+1, abs(f_original(x_k_next))
            
        x_k = x_k_next
        k += 1
        
    return x_k, k, abs(f_original(x_k))

# Função de iteração deduzida
def phi_exemplo(x):
    return x/2 + 1/x

raiz_pf, iter_pf, erro_pf = ponto_fixo(phi_exemplo, x0=1.5, f_original=f_exemplo, tol=1e-6)
raiz_pf, iter_pf, erro_pf
```

## Método de Newton-Raphson

O Método de Newton-Raphson é um método aberto incrivelmente poderoso e amplamente utilizado para encontrar raízes. Ele melhora a iterativa explorando informações da inclinação (derivada) da função no ponto atual.

Ao invés de conectar dois limites através de uma reta secante, o método traça uma reta **tangente** ao gráfico de $f(x)$ em $(x_k, f(x_k))$. A interseção dessa reta com o eixo $x$ produz a nossa próxima estimativa $x_{k+1}$.

## Componentes do Algoritmo

Este método pode ser implementado como um caso especial do Método do Ponto Fixo. A mecânica de iterar valores é exatamente a mesma, mas a função $\Phi(x)$ assume uma forma geométrica e algebricamente embasada nas derivadas.

- **$f'(x)$**: Derivada da função principal avaliada a cada passo.
- **$\Phi(x)$**: A função de iteração do método de Newton é desenhada como:
  
  $$\Phi(x) = x - \frac{f(x)}{f'(x)}$$

- **$x_0$**: Chute inicial.
- **$x_{k+1}$**: O próximo valor, gerado por $\Phi(x_k)$.

- **Critérios de Parada**: Como todo método base, avalia a diferença sucessiva $|x_k - x_{k-1}| < \epsilon$ ou a tolerância na imagem $|f(x_k)| < \epsilon$.

## Características do Método
- **Convergência Local**: Diferente dos métodos baseados em quebra de intervalo, Newton-Raphson exige um "bom palpite" $x_0$. Se tentarmos avaliar onde $f'(x_k)$ aproxima-se de zero (tangente horizontal), o cálculo pode ser lançado para extremos divergentes.
- **Velocidade Superior**: Se o palpite for seguro e as propriedades viáveis, a grande vantagem de Newton-Raphson se exibe: convergência **Quadrática**. A quantidade de casas decimais precisas praticamente dobra a cada iteração!


### Experimento

Buscando as raízes de $f(x) = x^2 - 2$ a partir do palpite estipulado $x_0 = 1.5$, usaremos exatamente a função iterativa de Ponto Fixo já declarada acima, uma vez que a lógica é a mesma.

Apenas definiremos a nossa base para formar a função iteradora: a derivada será $f'(x) = 2x$ e a função $\Phi_{NR}(x)$ será montada dinamicamente chamando a função original e sua derivada.


```python
# Derivada analítica de f(x) = x^2 - 2
def fd_exemplo(x):
    return 2 * x

# Função de iteração para o método de Newton-Raphson
def phi_newton(x):
    return x - (f_exemplo(x) / fd_exemplo(x))

# Invocando o Método de Newton reaproveitando a função "ponto_fixo"
raiz_nr, iter_nr, erro_nr = ponto_fixo(phi_newton, x0=1.5, f_original=f_exemplo, tol=1e-6)
raiz_nr, iter_nr, erro_nr
```

## Método da Secante

O Método de Newton-Raphson é extremamente rápido, porém exige o conhecimento ou o cálculo da derivada $f'(x)$. Em muitas situações práticas, derivar a função analiticamente é custoso ou até impossível.

O Método da Secante contorna este problema aproximando a reta tangente de Newton por uma **reta secante** que cruza a função em dois pontos avaliados anteriormente, $(x_{k-1}, f(x_{k-1}))$ e $(x_k, f(x_k))$.

## Componentes do Algoritmo

Este método é considerado parcialmente aberto. Diferente do confinamento da Bissecção, ele de fato parte por avaliar a interseção da reta, mas ao invés de usar a derivada exata no ponto, ele usa uma diferença finita.

- **$x_0$ e $x_1$**: O método **exige dois palpites iniciais**. Diferente da Falsa Posição, eles não precisam necessariamente encapsular a raiz apresentando sinais opostos em suas imagens.
- **$x_{k+1}$**: O próximo valor cruza a partir da interpolação linear dos pontos passados:
  
  $$x_{k+1} = x_k - f(x_k) \cdot \frac{x_k - x_{k-1}}{f(x_k) - f(x_{k-1})}$$

- **Critérios de Parada**: Como os outros métodos abertos, interrompe-se o ciclo se a proximidade atende tolerância $|x_k - x_{k-1}| < \epsilon$ ou o erro da variação na imagem $|f(x_k)| < \epsilon$.

## Características do Método
- **Convergência**: Possui as mesmas vulnerabilidades de Newton. Uma inclinação que aproxima de zero (linha paralela ao eixo $x$) pode jogar a nova estimativa muito para longe. Requer estimativas iniciais razoáveis.
- **Velocidade**: Mais lento que Newton-Raphson, porém superlinear em relação à Bissecção. Sua taxa de ordem de convergência é equivalente à razão áurea, $p \approx 1.618$!


### Experimento

Buscando as raízes de $f(x) = x^2 - 2$, usaremos as estimativas iniciais $x_0 = 1.0$ e $x_1 = 2.0$.

Como a Secante avalia sempre informações iterativas de dois instantes no tempo ($k$ e $k-1$), diferentemente de Newton-Raphson que avalia apenas do instante atual, nós **não** poderemos reaproveitar o algoritmo de `ponto_fixo` construído porque ele tem "memória de apenas 1 ponto". Sendo assim, construímos uma iterativa própria.


```python
def secante(f, x0, x1, tol=1e-5, max_iter=100):
    k = 0
    if abs(f(x0)) < tol:
        return x0, k, abs(f(x0))
    if abs(f(x1)) < tol:
        return x1, k, abs(f(x1))
    
    while k < max_iter:
        # Avanço usando a secante
        x_next = x1 - f(x1) * ((x1 - x0) / (f(x1) - f(x0)))
        
        if abs(f(x_next)) < tol or abs(x_next - x1) < tol:
            return x_next, k+1, abs(f(x_next))
            
        # Rotacionando os valores no tempo: atraso de 1 ponto
        x0 = x1
        x1 = x_next
        k += 1
        
    return x1, k, abs(f(x1))

# Invocando o Método da Secante
raiz_sec, iter_sec, erro_sec = secante(f_exemplo, 1.0, 2.0, tol=1e-6)
raiz_sec, iter_sec, erro_sec
```

## Método de Newton Modificado

Se por um lado o Método da Secante evita o cálculo avaliativo constante da derivada substituindo-a por uma diferença finita aproximada a cada passo temporal, o **Método de Newton Modificado** propõe uma alternativa diferente para diminuir o custo de derivar.

Nesta modificação, a inclinação usada para gerar a reta a cada iteração passa a ser considerada **constante**. Nós derivamos a função apenas em torno do ponto fixo inicial $x_0$, fixamos $f'(x_0)$ e montamos todas as estimativas sucessivas $x_{k+1}$ usando retas iterativas baseadas sempre com a mesma inclinação da época do palpite $x_0$.

## Componentes do Algoritmo

Este método, de forma surpreendente, cai como uma luva nos moldes clássicos do Método do Ponto Fixo. Basta utilizarmos aquele único primeiro palpite avaliado na derivada para predefinirmos a $\Phi(x)$.

- **$f'(x_0)$**: Derivada avaliada estritamente no chute inicial. Permanecerá inalterada no decorrer das iterações $k$.
- **$\Phi(x)$**: A função passo de reiterações assume o perfil linear:
  
  $$\Phi(x) = x - \frac{f(x)}{f'(x_0)}$$

## Características do Método
- **Ganho Computacional**: Reduz brutalmente o esforço avaliativo na derivada.
- **Velocidade**: Existe um contrapeso de troca! Ao remover o rastreio da curva por derivadas contemporâneas e fixar a derivada no passado remoto $x_0$, o método perde sua formidável convergência Quadrática e volta para uma taxa de convergência puramente **Linear** no limite assintótico.


### Experimento

Mais uma vez, buscando as raízes de $f(x) = x^2 - 2$ a partir do nosso palpite de $x_0 = 1.5$.

Podemos resgatar exatamente a nossa função implementada para o **Ponto Fixo**. 
A derivada estipulada de entrada $f'(x_0) = 2(1.5) = 3$ passará a embasar a mecânica, ou seja, $\Phi_{NM}(x) = x - \frac{f(x)}{3}$ será calculada com esse divisor cravado para o avanço dos passos.


```python
# Avaliando antecipadamente o custo caro de derivar apenas uma vez em x0=1.5!
df_x0 = fd_exemplo(1.5)

# Função de iteração de Ponto Fixo formatada para a modificação de Newton
def phi_newton_modificado(x):
    return x - (f_exemplo(x) / df_x0)

# Invocando as lógicas através do laço iterativo de ponto fixo
raiz_nm, iter_nm, erro_nm = ponto_fixo(phi_newton_modificado, x0=1.5, f_original=f_exemplo, tol=1e-6)
raiz_nm, iter_nm, erro_nm
```
