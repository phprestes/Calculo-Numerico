```python
import numpy as np
from scipy.differentiate import jacobian

def metodo_newton(F, x0, tol=0.05, max_iter=200):
    x_k = x0
    k = 0

    while k < max_iter:
        J = jacobian(F, x_k).df
        s_k = np.linalg.solve(J, -F(x_k))
        x_k_new = x_k + s_k

        if ((np.linalg.norm(x_k_new - x_k, np.inf) / np.linalg.norm(x_k_new, np.inf)) < tol):
            return x_k_new, k+1, (F(x_k_new))
        
        x_k = x_k_new
        k += 1
    
    return x_k, k, (F(x_k))

## f1(x) = x_0^2 + x_1^2 - 4; f2(x) = x_0 * x_1 - 1
def F(x):
    return np.array([x[0]**2 + x[1]**2 - 4, x[0] * x[1] - 1])
x0 = np.array([1.5, 0.5])

metodo_newton(F, x0)
```

```python
def metodo_newton_modificado(F, x0, tol=0.05, max_iter=200):
    x_k = x0
    k = 0
    J = jacobian(F, x0).df

    while k < max_iter:
        s_k = np.linalg.solve(J, -F(x_k))
        x_k_new = x_k + s_k

        if ((np.linalg.norm(x_k_new - x_k, np.inf) / np.linalg.norm(x_k_new, np.inf)) < tol):
            return x_k_new, k+1, (F(x_k_new))
        
        x_k = x_k_new
        k += 1
    
    return x_k, k, (F(x_k))

metodo_newton_modificado(F, x0)
```

Metodo de Newton Discreto

```python
def jacobiano_discreto(F, x_k, h=0.005):
    n = len(x_k)
    F_k = F(x_k)

    J = np.zeros((len(F_k), n))
    for j in range(n):
        x_pert = x_k.copy()
        x_pert[j] += h

        J[:, j] = (F(x_pert) - F_k) / h
    
    return J

def metodo_newton_discreto(F, x0, tol=0.05, max_iter=200):
    x_k = x0
    k = 0

    while k < max_iter:
        J = jacobiano_discreto(F, x_k)
        s_k = np.linalg.solve(J, -F(x_k))
        x_k_new = x_k + s_k

        if ((np.linalg.norm(x_k_new - x_k, np.inf) / np.linalg.norm(x_k_new, np.inf)) < tol):
            return x_k_new, k+1, (F(x_k_new))
        
        x_k = x_k_new
        k += 1
    
    return x_k, k, (F(x_k))

metodo_newton_discreto(F, x0)
```
